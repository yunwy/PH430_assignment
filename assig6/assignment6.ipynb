```python
# 2)
import numpy as np
import matplotlib.pyplot as plt
```

```python
K_O2 = 26
K_CO = K_O2/240
w_O2 = (0.21*760/K_O2)**3
print(f'w_O2={w_O2}')

h = 0.01
p_CO = np.arange(0, 20+h, h) # mmHg


def P_O2(p_CO):
    w_CO = (p_CO/K_CO)**(1.4)
    P = w_O2/(1 + w_O2 + w_CO) # Oxygen-bound probability

    return P
```

```python
P = P_O2(p_CO)

plt.axhline(y=0.5, color='r', linestyle='--', label='50%')
plt.plot(p_CO, P, label='$P(O_2)$')

plt.xlabel('$p_{CO}$')
plt.ylabel('$P(O_2)$')
plt.xlim((0, 20))
plt.ylim((0, 1))
plt.legend()

plt.show()
```

The trend of $P(O_2)$ is monotonically decreasing. If $P(O_2)=0.5$,
$$
\begin{gather*}
    0.5=\frac{w_{O_2}}{1 + w_{O_2} + w_{CO}}\\
    \therefore 1 + w_{CO} = w_{O_2}
\end{gather*}
$$
Therefore,
$$
\begin{align*}
    w_{CO}=\bigg(\frac{p_{CO}}{K_{CO}}\bigg)^{n_{CO}}&=w_{O_2}-1\\
    \frac{p_{CO}}{K_{CO}}&=(w_{O_2}-1)^{1/n_{CO}}\\
    p_{CO}&=K_{CO}(w_{O_2}-1)^{1/n_{CO}}\\
    &=\bigg(\frac{26~\mathrm{mmHg}}{240}\bigg)(231.3-1)^{1/1.4}
\end{align*}
$$


```python
p_CO_half = (26/240)*(w_O2-1)**(1/1.4)
print(f'p_CO={p_CO_half} mmHg')
```

Therefore, $P(O_2)$ falls below 50% after $p_{CO}\approx 5.27~\mathrm{mmHg}$.

$$
\begin{align*}
    P(O_2)&=P(CO)\\
    w_{O_2}&=w_{CO}\\
    \bigg(\frac{p_{CO}}{K_{CO}}\bigg)^{n_{CO}}&=w_{O_2}\\
    p_{CO}&=K_{CO}(w_{O_2})^{1/n_{CO}}
    &=\bigg(\frac{26~\mathrm{mmHg}}{240}\bigg)(231.3)^{1/1.4}
\end{align*}
$$


```python
p_CO_equal = (26/240)*(w_O2)**(1/1.4)
print(f'p_CO={p_CO_equal} mmHg')
```

Therefore, $P(O_2)$ and $P(CO)$ become equal at $p_{CO}\approx 5.29~\mathrm{mmHg}$. This value is very similar to $5.27~\mathrm{mmHg}$ because unbound state probability is very low. Independent Hill curves would be
$$
\begin{gather*}
    P(O_2)=\frac{w_{O_2}}{1+w_{O_2}},\quad P(CO)=\frac{w_{CO}}{1+w_{CO}}
\end{gather*}
$$
In this case, each ligand is treated as if the other were absent, so population increasing does not reduce the other's probability. Also, total probability is not fixed to 1, which is impossible for a shared binding siteTherefore, the common partition function $Z=1+w_{O_2}+w_{CO}$ is necessary for competition.
