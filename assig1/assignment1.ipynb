```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
```

```python
img = plt.imread("figure_02_03a.jpg")
profile = -img[500, 200:890, 0]
peaks, props = find_peaks(profile, distance=100)

plt.plot(profile)
plt.plot(peaks, 225*np.ones_like(peaks), 'o')

plt.xlabel('pixels')
plt.ylabel('intensity')

plt.show()
```

```python
distance = peaks[-1] - peaks[0]
distance_avg = int(np.round(distance/5))
print(f"10 µm is about {distance_avg} pixels")
```

```python
img2 = img.copy()
img2[840:870, 924:924+distance_avg, :] = 0
plt.imshow(img2)
plt.show()
```

```python
img3 = plt.imread("figure_02_03c.jpg").copy()
img3[800:800+distance_avg, 100:130, :] = 0

plt.imshow(img3)
plt.axis("off")
plt.text(140, 870, '10 µm')

plt.show()
```
